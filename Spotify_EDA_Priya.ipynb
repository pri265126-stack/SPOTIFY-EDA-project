```python
plt.figure(figsize=(8,5))


sns.histplot(df["popularity"], bins=20, color="green")
plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")

plt.show()
```

### Observation

The histogram shows that song popularity is not evenly distributed. Most songs have popularity between **20 and 60**, while very few songs have popularity above **80**. A large number of songs also have very low popularity, which indicates that only a small portion of songs are highly popular.


```python
plt.figure(figsize=(8,5))

sns.histplot(df["danceability"], bins=20, color="skyblue")

plt.title("Distribution of Danceability")
plt.xlabel("Danceability")
plt.ylabel("Number of Songs")

plt.show()
```

### Observation

The danceability values are mostly concentrated between **0.5 and 0.7**. This indicates that a large number of songs in the dataset have medium to high danceability. Very few songs have extremely low (below 0.2) or extremely high (above 0.9) danceability values.


```python
plt.figure(figsize=(8,5))

sns.histplot(df["energy"], bins=20, color="orange")

plt.title("Distribution of Energy")
plt.xlabel("Energy")
plt.ylabel("Number of Songs")

plt.show()
```

### Observation

The histogram shows that most songs have **high energy values between 0.6 and 1.0**. This indicates that the dataset contains a large number of energetic songs, while songs with very low energy values are comparatively fewer.


```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df["popularity"], color="violet")

plt.title("Box Plot of Popularity")
plt.xlabel("Popularity")

plt.show()
```

### Observation

The box plot shows that the popularity values are spread across a wide range. The median popularity is around **35**, and most songs have popularity between **18 and 50**. A few songs with popularity close to **100** appear as outliers, indicating very highly popular songs.


```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="danceability", y="popularity", data=df, alpha=0.5)

plt.title("Danceability vs Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")

plt.show()
```

### Observation

The scatter plot shows the relationship between danceability and popularity. Songs with higher danceability are found across different popularity levels, but there is no strong linear relationship between the two features. This suggests that danceability alone does not determine the popularity of a song.



```python
# Correlation matrix
plt.figure(figsize=(12,8))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Spotify Features")

plt.show()
```

### Observation

The correlation heatmap shows the relationship between different numerical features in the Spotify dataset. The strongest positive correlation is between **energy and loudness (0.76)**, while the strongest negative correlation is between **energy and acousticness (-0.73)**. Most other features have weak or moderate correlations with each other.


```python
# Top 10 genres in the dataset
top_genres = df["track_genre"].value_counts().head(10)

plt.figure(figsize=(10,5))

sns.barplot(x=top_genres.index, y=top_genres.values, palette="viridis")

plt.title("Top 10 Most Common Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The bar chart shows the top 10 genres present in the dataset. All top genres contain almost **1000 songs each**, which indicates that the dataset is balanced across different music genres. This helps avoid bias towards a single genre during analysis.


```python

top_artists = df["artists"].value_counts().head(10)

plt.figure(figsize=(10,5))

sns.barplot(x=top_artists.values, y=top_artists.index)

plt.title("Top 10 Artists with the Highest Number of Songs")
plt.xlabel("Number of Songs")
plt.ylabel("Artists")

plt.show()
```

### Observation

The bar chart shows the artists with the highest number of songs in the dataset. **The Beatles** have the highest number of songs among the top artists, followed by **George Jones** and **Stevie Wonder**. This indicates that these artists are more frequently represented in the dataset.


```python


top_songs = df.sort_values(by="popularity", ascending=False).head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x="popularity",
    y="track_name",
    data=top_songs,
    palette="magma"
)

plt.title("Top 10 Most Popular Songs")
plt.xlabel("Popularity")
plt.ylabel("Track Name")

plt.show()
```

### Observation

The bar chart displays the songs with the highest popularity scores in the dataset. Songs such as **Unholy**, **Quevedo: Bzrp Music Sessions, Vol. 52**, **La Bachata**, and **I'm Good (Blue)** have popularity values close to **100**, making them the most popular tracks in the dataset.


```python


genre_popularity = (
    df.groupby("track_genre")["popularity"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=genre_popularity.values,
    y=genre_popularity.index,
    hue=genre_popularity.index,
    palette="crest",
    legend=False
)

plt.title("Top 10 Genres by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Genre")

plt.show()
```

### Observation

The bar chart shows the average popularity of different music genres. **Pop-film** has the highest average popularity, followed by **K-pop** and **Chill**. Genres like **Pop** and **Sertanejo** are also among the top 10 popular genres, with average popularity close to 48–50.
